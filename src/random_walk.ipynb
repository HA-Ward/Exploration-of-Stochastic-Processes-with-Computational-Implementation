```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def random_walk(start, steps, p):
    q = 1-p
    step_set = np.array([-1,1])
    probs = np.array([q,p])
    random_steps = np.random.choice(step_set, steps, p=probs)

    return np.concatenate([[start], [start] + np.cumsum(random_steps)])
```

```python
#simple random symetric walk

path = random_walk(0,300,0.5)

plt.plot(path)
plt.title("Simple Random Symetric Walk")
plt.xlabel("Steps")
plt.ylabel("Position")
```

```python
#symetric multiple path

paths = 10000
final_positions = []

for i in range(paths):
    path = random_walk(0,300,0.5)
    final_positions.append(path[-1])
    plt.plot(path)

print("mean:", np.mean(final_positions))
print("var:", np.var(final_positions))

plt.title("Simple Random Symetric Walk (Multiple Paths)")
plt.xlabel("Steps")
plt.ylabel("Position")
plt.show()

plt.hist(final_positions, bins=80)
plt.title("Histogram of Final Positions")
plt.xlabel("Final Position")
```

The theoretical value for expectation and varience is 0 and 300
